```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LogisticRegression
```

```python
# Study Hours
X = np.array([1, 2, 3, 4, 5, 6, 7, 8])

# Pass / Fail
y = np.array([0, 0, 0, 0, 1, 1, 1, 1])

X = X.reshape(-1, 1)
```

```python
model = LogisticRegression()

model.fit(X, y)
```

```python
predictions = model.predict(X)

print("Predictions:")
print(predictions)


probs = model.predict_proba(X)

print("\nProbabilities:")
print(probs)


print("\nWeight (Coefficient):")
print(model.coef_)

print("\nBias (Intercept):")
print(model.intercept_)


w = model.coef_[0][0]
b = model.intercept_[0]

decision_boundary = -b / w

print("\nDecision Boundary =", decision_boundary)
```

```python
x_values = np.linspace(0, 10, 100).reshape(-1, 1)

y_prob = model.predict_proba(x_values)[:, 1]

plt.figure(figsize=(8,5))

plt.plot(x_values, y_prob)

plt.scatter(X, y)

plt.axvline(
    x=decision_boundary,
    linestyle='--'
)

plt.xlabel("Study Hours")
plt.ylabel("Probability of Passing")

plt.title("Logistic Regression using scikit-learn")

plt.show()
```
