```python
import numpy as np
import matplotlib.pyplot as plt


class LinearRegressionScratch:

    def __init__(self):
        self.w = 0
        self.b = 0

    def predict(self, x):
        return self.w * x + self.b

    def compute_cost(self, x, y):

        m = len(x)

        predictions = self.predict(x)

        cost = (1 / (2 * m)) * np.sum((predictions - y) ** 2)

        return cost

    def fit(self, x, y, learning_rate=0.01, epochs=1000000):

        m = len(x)

        cost_history = []

        for epoch in range(epochs):

            predictions = self.predict(x)

            dw = (1 / m) * np.sum((predictions - y) * x)
            db = (1 / m) * np.sum(predictions - y)

            self.w = self.w - learning_rate * dw
            self.b = self.b - learning_rate * db

            cost = self.compute_cost(x, y)

            cost_history.append(cost)

            if epoch % 100 == 0:
                print(f"Epoch {epoch}, Cost: {cost:.4f}")

        return cost_history


x = np.array([1, 2, 3, 4, 5, 6, 7, 8])
y = np.array([35, 40, 50, 55, 65, 70, 80, 90])

model = LinearRegressionScratch()

costs = model.fit(x, y)

print("Weight:", model.w)
print("Bias:", model.b)

prediction = model.predict(13)

print("Prediction:", prediction)

plt.scatter(x, y)

plt.plot(x, model.predict(x))

plt.show()

plt.plot(costs)

plt.title("Cost Reduction")

plt.show()
```

```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler


class LinearReg:

    def __init__(self, learning_rate=0.01, epochs=10000):

        self.learning_rate = learning_rate
        self.epochs = epochs

        self.w = 0
        self.b = 0

        self.cost_history = []

    # Prediction function
    def predict(self, x):

        return self.w * x + self.b

    # Cost function
    def compute_cost(self, x, y):

        m = len(x)

        predictions = self.predict(x)

        cost = (1 / (2 * m)) * np.sum((predictions - y) ** 2)

        return cost

    # Training
    def fit(self, x, y):

        # RESET parameters every training
        self.w = 0
        self.b = 0

        m = len(x)

        for epoch in range(self.epochs):

            predictions = self.predict(x)

            # Gradients
            dw = (1 / m) * np.sum((predictions - y) * x)
            db = (1 / m) * np.sum(predictions - y)

            # Update
            self.w = self.w - self.learning_rate * dw
            self.b = self.b - self.learning_rate * db

            # Save cost
            cost = self.compute_cost(x, y)
            self.cost_history.append(cost)

            # Print progress
            if epoch % 100 == 0:
                print(f"Epoch {epoch}, Cost: {cost:.4f}")


# ---------------- DATA ----------------

x = np.array([23, 26, 30, 34, 43, 48, 52, 57, 58])
y = np.array([651, 762, 856, 1063, 1190, 1298, 1421, 1440, 1518])


# ---------------- SCALING ----------------

x_scaler = MinMaxScaler()
y_scaler = MinMaxScaler()

scaled_x = x_scaler.fit_transform(x.reshape(-1, 1)).flatten()
scaled_y = y_scaler.fit_transform(y.reshape(-1, 1)).flatten()


# ---------------- MODEL ----------------

model = LinearReg(learning_rate=0.01, epochs=10000)

model.fit(scaled_x, scaled_y)


# ---------------- PREDICTION ----------------

# Predict first value
pred_scaled = model.predict(scaled_x[0])

# Convert back to original scale
pred_original = y_scaler.inverse_transform([[pred_scaled]])

print("\nScaled Prediction:", pred_scaled)
print("Original Prediction:", pred_original[0][0])


# ---------------- FINAL PARAMETERS ----------------

print("\nWeight:", model.w)
print("Bias:", model.b)


# ---------------- PLOT ----------------

import matplotlib.pyplot as plt

# Plot original data
plt.scatter(x, y)

# Predict all values
all_preds_scaled = model.predict(scaled_x)

# Convert predictions back
all_preds_original = y_scaler.inverse_transform(
    all_preds_scaled.reshape(-1, 1)
).flatten()

# Plot regression line
plt.plot(x, all_preds_original)

plt.title("Linear Regression")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()


# ---------------- COST GRAPH ----------------

plt.plot(model.cost_history)

plt.title("Cost Reduction")
plt.xlabel("Epochs")
plt.ylabel("Cost")

plt.show()
```

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

# Nonlinear data
X = np.linspace(-5, 5, 200)
y = X**2

# NN
model = tf.keras.Sequential([
    tf.keras.layers.Dense(
        8,
        activation='relu',
        input_shape=(1,)
    ),
    tf.keras.layers.Dense(1)
])

model.compile(
    optimizer='adam',
    loss='mse'
)

model.fit(
    X,
    y,
    epochs=500,
    verbose=0
)

y_pred = model.predict(X)

plt.scatter(X, y, s=10)
plt.plot(X, y_pred, linewidth=3,c='r')

plt.show()
```
