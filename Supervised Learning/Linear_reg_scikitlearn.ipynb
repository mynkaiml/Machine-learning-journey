```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)

y = np.array([35, 45, 50, 60, 70])
```

```python
model = LinearRegression()
```

```python
model.fit(X,y)
```

```python
pred = model.predict(X)
```

```python
model.coef_[0]
model.intercept_
```

```python
new_hour = [[6]]

predicted_marks = model.predict(new_hour)

predicted_marks[0]
```

```python
plt.scatter(X, y, label="Actual Data")
plt.plot(X, pred, label="Regression Line")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.legend()
plt.show()
```
