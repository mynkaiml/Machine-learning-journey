```python
import numpy as np

w = np.array([2, 3, 4])
b = 4
x = np.array([1, 2, 3])

n = len(x)

f = 0
for i in range(n):
    f += w[i] * x[i]
f += b

f
```

```python
import numpy as np
import time

start = time.time()

w = np.array([2, 3, 4])
b = 4
x = np.array([1, 2, 3])

n = len(x)

f = 0
for i in range(n):
    f += w[i] * x[i]

f += b

end = time.time()

print("Output:", f)
print("Execution Time:", end - start, "seconds")
```

```python
import numpy as np
import time

w = np.array([2, 3, 4])
x = np.array([1, 2, 3])
b = 4

start = time.perf_counter()

f = np.dot(w, x) + b

end = time.perf_counter()

execution_time = end - start

print("Output:", f)
print("Execution Time:", execution_time, "seconds")
```
